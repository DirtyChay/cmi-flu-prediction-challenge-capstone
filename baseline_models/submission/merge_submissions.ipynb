# Merge Baseline Submissions
Finds all individual task CSVs (`task_4_1.csv` … `task_4_6.csv`) in the current folder and merges them into a single `final_submission.csv`.
Tasks 4.7–4.10 are not produced by the baseline notebooks and are filled with 0.

```python
import glob
import re
from functools import reduce

import pandas as pd
```

```python
# Find all task CSVs in the current (submission) directory
csv_files = glob.glob('task_*.csv')
print(f"Found {len(csv_files)} files to merge: {sorted(csv_files)}\n")

if csv_files:
    # Read each CSV and merge on Participant_ID (outer join handles any missing participants)
    dfs = [pd.read_csv(f) for f in csv_files]
    final_df = reduce(lambda left, right: pd.merge(left, right, on='Participant_ID', how='outer'), dfs)

    # Tasks 4.7-4.10 are not produced by baseline notebooks — fill with 0
    for col in ['Task_4.7', 'Task_4.8', 'Task_4.9', 'Task_4.10']:
        final_df[col] = 0


    # Sort columns: Participant_ID first, then numerically by task number
    def sort_cols(col_name):
        if col_name == 'Participant_ID':
            return -1
        match = re.search(r'4[._](\d+)', col_name)
        return int(match.group(1)) if match else 999


    final_df = final_df[sorted(final_df.columns, key=sort_cols)]

    final_df.to_csv('final_submission.csv', index=False)
    print("Saved to 'final_submission.csv'")
    display(final_df)
else:
    print("No 'task_*.csv' files found in the current directory.")
```
