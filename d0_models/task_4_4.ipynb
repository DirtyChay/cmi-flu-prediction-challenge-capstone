```python
challenge_data = pd.read_csv(CHALLENGE_DATA_PATH)
```

```python
D0_cols = [f'HAI_{strain}_d0' for strain in VACCINE_STRAINS]
prediction_df = challenge_data[['participant_id']].copy()
prediction_df['Task_4.4'] = np.exp2(challenge_data[D0_cols].mean(axis=1, skipna=True))
prediction_df.rename(columns={'participant_id': 'Participant_ID'}, inplace=True)
prediction_df
```
