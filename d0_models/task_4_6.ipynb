# Task 4.6

**4.6 Predict antibody breadth - percentage of variants with HAI >= 40 (D28)**
* Training Data: Demographics + Day 0 + Day 7 innate
* Assay: HAI / Measure: Day 0 baseline / Metric: Spearman correlation

---

**Prediction rule:** assume no change from Day 0 to Day 28.

**Output scale:** HAI columns are log2-transformed in the cleaned data. CSVs use `np.exp2` for raw titer or raw geometric mean outputs, except Task 4.6 which is already a fraction.
* Full description: Percentage of variants with HAI >= 40 at Day 28


```python
import os
import numpy as np
import pandas as pd
```

```python
USE_GOOGLE_COLAB = False
HAI_THRESHOLD = 40

if USE_GOOGLE_COLAB:
    from google.colab import drive

    drive.mount('/content/drive')
    DATA_PATH = os.path.join(os.getcwd(), 'drive', 'MyDrive', 'DSE_260A', 'data', 'cleaned_data')
else:
    DATA_PATH = '../cleaned_data'

SUBMISSION_PATH = 'submission/task_4_6.csv'
CHALLENGE_DATA_PATH = '../cleaned_data/challenge_combined.csv'
```

```python
challenge_data = pd.read_csv(CHALLENGE_DATA_PATH)
```

# Predicting no change in HAI

predicted = baseline measurement at day 0


```python
D0_cols = [c for c in challenge_data.columns if c.startswith('HAI_') and c.endswith('_d0')]
threshold_log2 = np.log2(HAI_THRESHOLD)
d0_block = challenge_data[D0_cols]
prediction_df = challenge_data[['participant_id']].copy()
prediction_df['Task_4.6'] = (d0_block >= threshold_log2).sum(axis=1) / d0_block.notna().sum(axis=1)
prediction_df.rename(columns={'participant_id': 'Participant_ID'}, inplace=True)
prediction_df
```

```python
prediction_df.to_csv(SUBMISSION_PATH, index=False)
```
